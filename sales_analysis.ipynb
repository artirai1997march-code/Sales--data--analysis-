# Sales Data Analysis

Beginner-level analysis of 20 sales transactions using Python, Pandas, and Matplotlib.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 1. Load the dataset

```python
df = pd.read_csv('sales_data.csv')
df.head()
```

## 2. Understand the data

```python
print('Rows:', len(df))
print('Columns:', len(df.columns))
df.info()
```

```python
df.describe(include='all')
```

## 3. Check for missing values

```python
df.isnull().sum()
```

## 4. Overall sales analysis

```python
total_sales = df['Sales'].sum()
average_sales = df['Sales'].mean()
minimum_sales = df['Sales'].min()
maximum_sales = df['Sales'].max()
total_quantity = df['Quantity'].sum()

print(f'Total Sales: ₹{total_sales:,.0f}')
print(f'Average Sales: ₹{average_sales:,.0f}')
print(f'Minimum Sale: ₹{minimum_sales:,.0f}')
print(f'Maximum Sale: ₹{maximum_sales:,.0f}')
print(f'Total Quantity Sold: {total_quantity}')
```

## 5. Sales by product

```python
product_sales = df.groupby('Product')['Sales'].sum().sort_values(ascending=False)
product_sales
```

```python
product_sales.plot(kind='bar', title='Total Sales by Product')
plt.xlabel('Product')
plt.ylabel('Sales (₹)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 6. Quantity sold by product

```python
df.groupby('Product')['Quantity'].sum().sort_values(ascending=False)
```

## 7. Sales by region

```python
region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)
region_sales
```

```python
region_sales.plot(kind='bar', title='Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Sales (₹)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 8. Sales trend

```python
df['Date'] = pd.to_datetime(df['Date'])
daily_sales = df.groupby('Date')['Sales'].sum()
daily_sales
```

```python
daily_sales.plot(kind='line', marker='o', title='Daily Sales Trend')
plt.xlabel('Date')
plt.ylabel('Sales (₹)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 9. Key findings

```python
top_product = product_sales.idxmax()
top_region = region_sales.idxmax()
print(f'Top product by sales: {top_product} (₹{product_sales.max():,.0f})')
print(f'Top region by sales: {top_region} (₹{region_sales.max():,.0f})')
```
